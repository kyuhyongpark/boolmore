```python
from pathlib import Path
```

```python
folder = Path("./target_models/")
```

```python
def read_aeon(path):
    edges = {}

    with open(path) as f:
        for line in f:
            line = line.strip()

            if not line or line.startswith("#") or line.startswith("$"):
                continue

            regulator, edge_type, target = line.split()

            regulator = regulator.removeprefix("v_")
            target = target.removeprefix("v_")

            edges[(regulator, target)] = edge_type

    return edges


def compatible(original, inferred):
    if original == "-?":
        return True
    return original == inferred
```

```python
for original_file in sorted(folder.glob("[0-9][0-9][0-9].aeon")):

    inferred_file = folder / f"{original_file.stem}_inferred.aeon"

    if not inferred_file.exists():
        print(f"Missing {inferred_file.name}")
        continue

    original = read_aeon(original_file)
    inferred = read_aeon(inferred_file)

    only_original = set(original) - set(inferred)
    only_inferred = set(inferred) - set(original)

    disagreements = []

    for edge in set(original) & set(inferred):
        if not compatible(original[edge], inferred[edge]):
            disagreements.append(
                (edge[0], edge[1], original[edge], inferred[edge])
            )

    print(f"\n=== {original_file.stem} ===")

    if only_original:
        print("Only in original:")
        for r, t in sorted(only_original):
            print(f"  {r} {original[(r, t)]} {t}")

    if only_inferred:
        print("Only in inferred:")
        for r, t in sorted(only_inferred):
            print(f"  {r} {inferred[(r, t)]} {t}")

    if disagreements:
        print("Edge type disagreements:")
        for r, t, o, i in sorted(disagreements):
            print(f"  {r} -> {t}: {o} vs {i}")

    if not (only_original or only_inferred or disagreements):
        print("Perfect match.")
```
