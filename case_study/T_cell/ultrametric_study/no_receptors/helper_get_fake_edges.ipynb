```python
import pandas as pd
import random

input_file = "../../confidence_rank_ultrametric.csv"

add_probability = 1.0

output_file = "fake_edges_no_receptors.csv"
```

```python
df = pd.read_csv(input_file)

df.head()
```

```python
regulators = set(df["Regulator"]) | set(df["Target"])
targets = set(df["Target"]) - set(["IKB", "IL7R", "CD28", "TCR", "NOTCH1", "IFNAR", "IFNBR", "IL15R", "IL27R", "IL29R", "IL36R"])

n_possible = len(regulators) * len(targets)
n_existing = df[["Regulator", "Target"]].drop_duplicates().shape[0]

print(f"Number of regulators: {len(regulators)}")
print(f"Number of targets: {len(targets)}")
print(f"Possible regulator-target pairs: {n_possible}")
print(f"Existing regulator-target pairs: {n_existing}")
```

```python
existing_pairs = set(zip(df["Regulator"], df["Target"]))

new_rows = []

for regulator in regulators:
    for target in targets:
        if (regulator, target) not in existing_pairs:
            if random.random() > add_probability:
                continue
            new_rows.append({
                "Regulator": regulator,
                "Target": target,
                "Sign": "unknown",
                "Model": "Fake",
                "Confidence Rank": 6,
                "Constraint": "",
                "Direct": ""
            })

new_df = pd.DataFrame(new_rows, columns=df.columns)

print(f"Created {len(new_df)} new rows.")
```

```python
output_df = pd.concat([df, new_df], ignore_index=True)

output_df.to_csv(output_file, index=False)

print(f"Saved to: {output_file}")
print(f"Total rows: {len(output_df)}")
```
