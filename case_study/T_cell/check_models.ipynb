```python
import os

from pyboolnet.external.bnet2primes import bnet_text2primes

from pyboolnet.interaction_graphs import primes2igraph
```

```python
MODELS_DIR = "./target_models"

json_file = "./Tcell_config.json"
```

```python
# read every file that ends with .bnet in the current directory
edges_files = [f for f in os.listdir(MODELS_DIR) if f.endswith('edges.txt')]
edges_files.sort()
print(edges_files)
```

```python
edges_dict = {}
nodes_dict = {}
for f in edges_files:
    with open(MODELS_DIR + '/' + f, 'r') as file:
        
        print("importing", f)
        edges_text = file.read()
        
        edges = []
        nodes = set()
        for line in edges_text.split("\n"):
            if line.strip() == "":
                continue
            edge = line.split(" ")
            edges.append(edge)
            nodes.add(edge[0])
            nodes.add(edge[2])

        edges_dict[f.removesuffix("_edges.txt")] = edges
        nodes_dict[f.removesuffix("_edges.txt")] = sorted(nodes)

        print(f"\tnumber of nodes: {len(nodes)}")
        print(f"\tnumber of edges: {len(edges)}")
```

```python
print(edges_dict["P18_BBM66"])
print(nodes_dict["P18_BBM66"])
```

```python
# read every file that ends with .bnet in the current directory
bnet_files = [f for f in os.listdir(MODELS_DIR) if f.endswith('.bnet')]
bnet_files.sort()
print(bnet_files)
```

```python
primes_dict = {}
for f in bnet_files:
    with open(MODELS_DIR + '/' + f, 'r') as file:
        
        print("importing", f)
        bnet_text = file.read()
        primes = bnet_text2primes(bnet_text)

        n = len(primes)
        sources = []
        constants = []
        for node in primes:
            if primes[node] == [[{node:0}],[{node:1}]]:
                sources.append(node)
            elif primes[node] == [[{}],[]]:
                constants.append(node)
            elif primes[node] == [[],[{}]]:
                constants.append(node)

        print(f"\tnumber of nodes: {n}")
        print(f"\tnumber of source nodes: {len(sources)}")
        print(f"\tnumber of constant nodes: {len(constants)}")

        primes_dict[f.removesuffix(".bnet")] = primes
```

```python
primes_dict["N10_BBM155"].keys()
```

```python
for model in primes_dict:
    print(model)
    
    nodes_from_bnet = primes_dict[model].keys()
    nodes_from_edges = nodes_dict[model]

    print("\tnodes only in AEON")
    for node in nodes_from_edges:
        if node not in nodes_from_bnet:
            print("\t",node)
    
    print("\tnodes only in bnet")
    for node in nodes_from_bnet:
        if node not in nodes_from_edges:
            print("\t",node)
```

```python
ig_dict = {}
for model in primes_dict:
    ig = primes2igraph(primes_dict[model])
    ig_dict[model] = ig
```

```python
for model in ig_dict:
    print(model)
    print("number of nodes:", len(ig_dict[model].nodes()))
    print("number of edges:", len(ig_dict[model].edges()))

    self_edges = 0
    sources = 0
    for edge in ig_dict[model].edges():
        if edge[0] == edge[1]:
            # print(edge)
            self_edges += 1
            node = edge[0]
            if primes_dict[model][node] == [[{node:0}],[{node:1}]]:
                sources += 1

    print("number of self edges:", self_edges)
    print("number of source nodes:", sources)
    print("number of edges without source:", len(ig_dict[model].edges()) - sources)
    print("number of self edges without source:", self_edges - sources)
    print()
```

```python
for model in ig_dict:
    print(model)
    edges_from_bnet = ig_dict[model].edges()
    edges_from_bbm = edges_dict[model]

    print("\tedges only in AEON")
    for edge in edges_from_bbm:
        if (edge[0], edge[2]) not in edges_from_bnet:
            print("\t",edge)
    
    print("\tedges only in bnet")
    for edge in edges_from_bnet:
        if [edge[0], '->', edge[1]] in edges_from_bbm:
            continue
        elif [edge[0], '-|', edge[1]] in edges_from_bbm:
            continue
        elif [edge[0], '-|?', edge[1]] in edges_from_bbm:
            continue
        elif [edge[0], '-?', edge[1]] in edges_from_bbm:
            continue
        else:
            node = edge[0]
            if node == edge[1] and primes_dict[model][node] == [[{node:0}],[{node:1}]]:
                continue
            else:
                print("\t",edge)
    print()
```
