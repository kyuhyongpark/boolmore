```python
import pickle
import json
import cProfile
import pstats
import time

from boolmore.core.model import Model
from boolmore.algo.mutation import mutate_rr_constraint
```

```python
PRIME_CACHE = "Tcell_primes_ultrametric.pkl"

json_file = "Tcell_config.json"
```

```python
with open(PRIME_CACHE, "rb") as f:
    primes = pickle.load(f)
print("Loaded primes from cache.")
```

```python
f = open(json_file)
json_dict = json.load(f)

CONSTRAINTS = json_dict["constraints"]
EDGE_POOL = json_dict["edge_pool"]
```

```python
base_model = Model.import_model(primes, constraints=CONSTRAINTS,edge_pool=EDGE_POOL)
```

```python
def main_function():
    new_model = base_model.mutate(1, 0.5, 0.5)

def main_function():
    for node in base_model.rr_dict:
        if node != "IFNG": continue
        print(node)
        
        start_time = time.perf_counter()
        # get mutated_rr from rr
        mutated_rr, modified = mutate_rr_constraint(
            regulators=base_model.regulators_dict[node],
            rr=base_model.rr_dict[node],
            base_rr=base_model.rr_dict[node],
            constraints=base_model.constraints,
            node=node,
            probability=0.5,
            bias=0.5
            )
        end_time = time.perf_counter()
        print(f"Time taken: {end_time - start_time:.4f} seconds")
```

```python
# Profile only the targeted function block
with cProfile.Profile() as profiler:
    main_function()

# Format and print the results
stats = pstats.Stats(profiler)
stats.sort_stats(pstats.SortKey.CUMULATIVE)
stats.print_stats(10)  # Limit output to top 10 rows
```
