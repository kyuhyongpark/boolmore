```python
from pyboolnet.external.bnet2primes import bnet_text2primes

bnet = """
A, A
B, B
C, A & B
D, D & B
E, !E
"""

primes = bnet_text2primes(bnet)

source_nodes = []
for node in primes:
    if primes[node] == [[{node:0}],[{node:1}]]:
        source_nodes.append(node)
source_order = sorted(source_nodes)
print("Source nodes:", source_nodes)

sources_partial = {"B": 1}
print("Partial assignment:", sources_partial)

answer_sheet = {}
for mask in generate_source_masks(source_order, sources_partial):
    source_comb = mask_to_sources(mask, source_order)
    print("Source combination:", source_comb)

    phenotypes = get_mintr_for_source_comb(primes, source_comb)
    print("Phenotypes:", phenotypes)

    answer_sheet[mask] = phenotypes

print(f"Answer sheet with sources {source_order}:")
for mask, phenotypes in answer_sheet.items():
    print(f"{mask_to_str(mask, len(source_order))}: {phenotypes}")

print(answer_sheet)
```

```python
from pyboolnet.external.bnet2primes import bnet_text2primes

bnet = """
A, A
B, B
C, A & B
D, D & B
E, !E
"""

target_phenotype = {'C': 0, 'D': 1}
print("Target phenotype:", target_phenotype)

sources_partial = {"B": 1}
print("Partial assignment:", sources_partial)


primes = bnet_text2primes(bnet)

source_nodes = []
for node in primes:
    if primes[node] == [[{node:0}],[{node:1}]]:
        source_nodes.append(node)
source_order = sorted(source_nodes)
print("Source nodes:", source_nodes)

answer_sheet = {}
for mask in generate_source_masks(source_order, sources_partial):
    source_comb = mask_to_sources(mask, source_order)
    print("Source combination:", source_comb)

    phenotypes = get_mintr_for_source_comb(primes, source_comb)
    print("Phenotypes:", phenotypes)

    answer_sheet[mask] = phenotypes

    match = find_matching_phenotype({mask: phenotypes}, target_phenotype)
    if match is not None:
        print(f"Target phenotype {target_phenotype} exists in source combination {mask_to_str(mask, len(source_order))} as minimal trap space {match[1]}.")
        break

print(f"Answer sheet with sources {source_order}:")
for mask, phenotypes in answer_sheet.items():
    print(f"{mask_to_str(mask, len(source_order))}: {phenotypes}")
```
