```python
from pyboolnet.external.bnet2primes import bnet_text2primes
from pystablemotifs.format import primes2bnet

from boolmore.conversions import merge_rules
```

```python
# example should include constant nodes, source nodes, etc
# it should also make constants and source nodes become non-sources
bnet1 = """
A, 1
B, B
C, A & B
"""

bnet2 = """
X, 0
Y, Y
Z, X & Y
"""

bnet3 = """
A, X
X, A
B, Y
Y, Y
C, Y
"""
```

```python
model1_primes = bnet_text2primes(bnet1)
model2_primes = bnet_text2primes(bnet2)
model3_primes = bnet_text2primes(bnet3)
```

```python
merged_primes, regulators, _, _ = merge_rules([model1_primes, model2_primes, model3_primes])

print(merged_primes)
```

```python
merged_bnet = primes2bnet(merged_primes)
print(merged_bnet)
```

```python
for node in regulators:
    print(f"{node}: {regulators[node]}")
```
