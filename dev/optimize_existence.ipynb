```python
import time
import pickle
from pyboolnet.external.bnet2primes import bnet_text2primes
from pyboolnet.trap_spaces import compute_trapspaces_within_subspace
```

```python
sources_partial = {"IL7_e":0, "DLL1":0}


# Choose a phenotype that (ideally) does not exist for worst-case benchmarking
target_phenotype = {'TBET':0, 'IFNG':0, 'GATA3':0, 'IL4':0, 'FOXP3':0,
                    'TGFB':0, 'RORGT':0, 'IL17':0, 'BCL6':0, 'PU1':0,
                    'STAT3':0, 'IL5':0, 'IL9':0, 'IL13':0, 'IL21':0,
                    'IL22':0}

CACHE_FILE = "primes_cache.pkl"

with open(CACHE_FILE, "rb") as f:
    primes = pickle.load(f)
print("Loaded primes from cache.")
```

```python
source_nodes = [
    node
    for node in primes
    if primes[node] == [[{node: 0}], [{node: 1}]]
]
source_order = sorted(source_nodes)

n_free = len(source_order) - len(sources_partial)
n_total = 2 ** n_free

print(f"Source nodes      : {source_order}")
print(f"Partial assignment: {sources_partial}")
print(f"Target phenotype  : {target_phenotype}")
print(f"Free source nodes : {n_free}")
print(f"Total combinations: {n_total}")
print()

start = time.perf_counter()
last_print = start
processed = 0

subspace = sources_partial.copy()
subspace.update(target_phenotype)

for node in subspace:
    if node not in primes:
        raise ValueError(f"Node {node} not in primes.")

phenotypes = compute_trapspaces_within_subspace(primes,
                                                subspace = subspace,
                                                type_="max",
                                                max_output=1)

elapsed = time.perf_counter() - start

if phenotypes:
    
    print("\nMATCH FOUND")
    print(f"Trap space: {phenotypes[0]}")
    print(f"Elapsed            : {elapsed:.2f} s")
    
else:
    print("\nNO MATCH FOUND")
    print(f"Elapsed  : {elapsed:.2f} s")
```
