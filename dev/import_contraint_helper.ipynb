```python
import csv
from collections import defaultdict
```

```python
file_path = "confidence_rank_ultrametric.csv"
```

```python
def read_constraints_csv(path):
    result = {
        "regulates": defaultdict(set),
        "necessary": defaultdict(set),
        "group": defaultdict(set),
    }

    with open(path, newline="") as f:
        reader = csv.DictReader(f)

        for row in reader:
            constraint = row["Constraint"].strip()
            target = row["Target"].strip()
            regulator = row["Regulator"].strip()

            if not constraint:
                continue

            if constraint not in result:
                print(f"unexpected constraint {constraint}")
                continue  # ignore unexpected values

            result[constraint][target].add(regulator)

    # convert sets -> sorted lists
    final_result = {
        k: {t: sorted(list(regs)) for t, regs in sorted(v.items())}
        for k, v in result.items()
    }

    return final_result
```

```python
constraints = read_constraints_csv(file_path)
constraints
```
