```python
import os

import csv
```

```python
MODELS_DIR = "../case_study/T_cell/target_models"

csv_file = "./confidence_rank.csv"
```

```python
# read every file that ends with .bnet in the current directory
edges_files = [f for f in os.listdir(MODELS_DIR) if f.endswith('edges.txt')]
edges_files.sort()
print(edges_files)
```

```python
edges = []
unique_edges = set()
nodes = set()
for f in edges_files:
    with open(MODELS_DIR + '/' + f, 'r') as file:
        
        print("importing", f)
        edges_text = file.read()
        
        for line in edges_text.split("\n"):
            if line.strip() == "":
                continue
            edge = line.split(" ")
            edges.append((edge[0], edge[2], f.split("_")[0]))
            unique_edges.add((edge[0], edge[2]))
            nodes.add(edge[0])
            nodes.add(edge[2])

nodes = sorted(list(nodes))
unique_edges = sorted(list(unique_edges))

print()
print("number of nodes:", len(nodes))
print("number of edges:", len(edges))
print("number of unique edges:", len(unique_edges))
```

```python

# Initialize your collection containers
csv_edges = []
csv_unique_edges = set()

# Open and parse the data file safely
with open(csv_file, mode='r', newline='', encoding='utf-8') as file:
    reader = csv.DictReader(file)
    
    for row in reader:
        # 1. Create a 3-column tuple and append to the list
        edge_with_model = (row['Regulator'], row['Target'], row['Model'])
        if edge_with_model in csv_edges:
            print("duplicate edge:", edge_with_model)
        csv_edges.append(edge_with_model)
        
        # 2. Create a 2-column tuple and add to the set (automatically drops duplicates)
        edge = (row['Regulator'], row['Target'])
        csv_unique_edges.add(edge)

# Display your results
print("Number of edges:", len(csv_edges))
print("Number of unique edges:", len(csv_unique_edges))
```

```python
# compare two lists and sets
print("edges in AEON but not in csv file")
for edge in edges:
    if edge not in csv_edges:
        print(edge)

print("edges in csv file but not in AEON")
for edge in csv_edges:
    if edge not in edges:
        print(edge)

print("unique edges in AEON but not in csv file")
for edge in unique_edges:
    if edge not in csv_unique_edges:
        print(edge)

print("unique edges in csv file but not in AEON")
for edge in csv_unique_edges:
    if edge not in unique_edges:
        print(edge)
```
